```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

# For demonstration, here's a mock dataset:
#read csv
df = pd.read_csv('./data/dreps-registrations.csv')
print(df)

# Calculate net change
df['net_change'] = df['registrations'] - df['deregistrations']

# Set epoch as x-axis
x = df['epoch_no']

# Plot setup
plt.figure(figsize=(15, 5))

# Plot 1: Registrations
plt.subplot(1, 3, 1)
plt.plot(x, df['registrations'], marker='o', color='green')
plt.title('DRep Registrations per Epoch')
plt.xlabel('Epoch')
plt.ylabel('Registrations')
plt.grid(True)

# Plot 2: Deregistrations
plt.subplot(1, 3, 2)
plt.plot(x, df['deregistrations'], marker='o', color='red')
plt.title('DRep Deregistrations per Epoch')
plt.xlabel('Epoch')
plt.ylabel('Deregistrations')
plt.grid(True)

# Plot 3: Net Change
plt.subplot(1, 3, 3)
plt.plot(x, df['net_change'], marker='o', color='blue')
plt.axhline(0, linestyle='--', color='gray')
plt.title('Net Change in DReps per Epoch')
plt.xlabel('Epoch')
plt.ylabel('Net Change')
plt.grid(True)

# Layout
plt.tight_layout()
plt.show()
```
